Create these tensors:

2. A zeros tensor with shape `(2, 5)`.

4. A range tensor containing values from `0` to `9`.

Print each tensor, its shape, and its dtype.

````python
t1 = torch.rand((3, 4,))
t2 = torch.zeros((2, 5,))
t3 = torch.ones((4,))
t4 = torch.arange(0, 9)

print(t1, t1.shape, t1.dtype, t1.device)
print(t2, t2.shape, t2.dtype, t2.device)
print(t3, t3.shape, t3.dtype, t3.device)
print(t4, t4.shape, t4.dtype, t4.device)
````
